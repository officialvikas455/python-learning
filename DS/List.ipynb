### print all positive and negative elements


```python
l = [3,-1,4,-5,9]

pos = []
neg = []

for i in l :
  if i >= 0:
    pos.append(i)
  else:
    neg.append(i)
print(f"Positive = {pos}",f"Negative = {neg}")
```

## Find the greatest element and print its index

```python
a = [8,9,8,10,12]

largest = a[0]
index = 0
for i in range(len(a)):
    if a[i] > largest:
        largest = a[i]
        index = i
        print(largest, index)
```

```python
a = [8,9,8,10,12]
largest = a[0]
sec_largest = a[0]

for i in a:
    if i > largest:
        sec_largest = largest
        largest = i
    elif i > sec_largest:
        sec_largest = i

print(largest, sec_largest)
```

```python
a = [8,9,8,10,12]

for i in range(len(a)-1):
    if a[i] > a[i+1]:
        print("Your list is not sorted")
        break
    else:
        print(" sorted")
```
