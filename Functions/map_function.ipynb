```python
nums = [10,20,30,40]
result = list(map(lambda x : x+5, nums))
print(result)
```

```python
nums = ["10", "20", "30", "40"]

result = list(map(int, nums))

print(result)
```

```python
a = [1, 2, 3]
b = [10, 20, 30]

result = list(map(lambda x, y: x + y, a, b))

print(result)
```
