```python
def temp_change(temp_value, unit):
    if unit == 'C':
        temperature = (temp_value * 9/5) + 32
        print(f"The temperature for the given value {temp_value}°C is: {temperature}°F")
    elif unit == 'F':
        temperature = (temp_value - 32) * 5/9
        print(f"The temperature for the given value {temp_value}°F is: {temperature}°C")
    else:
        print("Invalid unit. Please use 'C' or 'F'.")

# Test the function
temp_change(7, 'C')
```

```python
def password_strength(password):
    # Check length (Should be at least 8 characters)
    if len(password) < 8:
        print(f"The entered password '{password}' is Weak (Too short)")
        return

    # Check for lowercase letters
    if not any(char.islower() for char in password):
        print(f"The entered password '{password}' is Weak (Missing lowercase letter)")
        return

    # Check for uppercase letters
    if not any(char.isupper() for char in password):
        print(f"The entered password '{password}' is Weak (Missing uppercase letter)")
        return

    # Check for numbers
    if not any(char.isdigit() for char in password):
        print(f"The entered password '{password}' is Weak (Missing a number)")
        return

    # Check for special characters
    if not any(not char.isalnum() for char in password):
        print(f"The entered password '{password}' is Weak (Missing a special character)")
        return

    # If all checks pass
    print(f"The entered password '{password}' is Strong!")

# Test the function
password_strength("Pssword123")
```

```python
def palindrome(string):
    string_value = string.lower().replace(" ", "")

    if string_value == string_value[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")


palindrome("ba")
```
