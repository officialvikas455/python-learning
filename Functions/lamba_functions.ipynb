## Lamba Functions are the one liner function defined by lamba function

## Lambda ek small anonymous function hota hai, jisme def likhne ki zarurat nahi hoti.

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"
```

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"
```

```python
nums = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x * x*x, nums))
print(squares)
```

```python
# Maximum of two numbers
maximum = lambda a, b: a if a > b else b

print(maximum(10, 20))
```

```python
check = lambda x:"Even" if x % 2 == 0 else "odd"
print(check(6))
```

```python
check = lambda x: "Positive" if x > 0 else "Negative"
print(check(-5))
```

```python
nums = [1, 2, 3, 4, 5, 6]

odd = list(filter(lambda x: x % 2 != 0, nums))

print(odd)
```

```python
def square(num):
    return num**2;
print(square(6))
```
